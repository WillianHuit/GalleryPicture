```
import toytree

class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None
        self.anterior = None


class ArbolBinario:
    def __init__(self):
        self.raiz = None

    def insertar(self, valor):
        if self.raiz is None:
            self.raiz = Nodo(valor)
        else:
            self._insertar_recursivo(self.raiz, valor)

    def _insertar_recursivo(self, nodo, valor):
        if valor < nodo.valor:
            if nodo.anterior is None:
                nodo.anterior = Nodo(valor)
            else:
                self._insertar_recursivo(nodo.anterior, valor)
        else:
            if nodo.siguiente is None:
                nodo.siguiente = Nodo(valor)
            else:
                self._insertar_recursivo(nodo.siguiente, valor)
    # ---------------------------------
    def inorden(self):
        return self._inorden_recursivo(self.raiz)
    def _inorden_recursivo(self, nodo):
        if nodo is None:
            return []
        return self._inorden_recursivo(nodo.anterior) + [nodo.valor] + self._inorden_recursivo(nodo.siguiente)
    # ---------------------------------
    def preorden(self):
        return self._preorden_recursivo(self.raiz)
    def _preorden_recursivo(self, nodo):
        if nodo is None:
            return []
        return [nodo.valor] + self._preorden_recursivo(nodo.anterior) + self._preorden_recursivo(nodo.siguiente)
    # ---------------------------------
    def postorden(self):
        return self._postorden_recursivo(self.raiz)
    def _postorden_recursivo(self, nodo):
        if nodo is None:
            return []
        return self._postorden_recursivo(nodo.anterior) + self._postorden_recursivo(nodo.siguiente) + [nodo.valor]

# Solicitud principal


arbol = ArbolBinario()

print("Ingrese un número, ingrese # cuando termine de ingresar números")
print("Debe ingresar al menos 10 números")
entrada = input()
contador = 0

while True:
    if entrada == "#":
        if contador < 10:
            print("Debe ingresar al menos 10 números antes de finalizar.")
        else:
            print("Fin de la entrada.")
            break
    else:
        try:
            numero = int(entrada)
            arbol.insertar(numero)
            contador += 1
            if contador >= 50:
                print("Se alcanzó el límite de 50 números.")
                break
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número o # para finalizar.")

    entrada = input(f"Ingrese el número #{contador + 1}: ")

print("Datos ingresados:", contador)


print("Recorridos del árbol binario:")
print("Inorden:", arbol.inorden())
print("Preorden:", arbol.preorden())
print("Postorden:", arbol.postorden())

print("Generando visualización del árbol...")
def construir_toytree(nodo):
    if nodo is None:
        return None
    tnode = toytree.Toytree(name=str(nodo.valor))
    tnode.add_child(construir_toytree(nodo.anterior))
    tnode.add_child(construir_toytree(nodo.siguiente))
    return tnode
toytree_arbol = construir_toytree(arbol.raiz)
toytree_arbol.draw("arbol_binario.png")
```
