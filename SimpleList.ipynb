```
class SimpleNode:
    def __init__(self, value):
        self.value = value
        self.next = None

class SimpleList:
    def __init__(self):
        self.head = None

    def isEmpty(self):
        return self.head is None

    def size(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def addFirst(self, value):
        new_node = SimpleNode(value)
        new_node.next = self.head
        self.head = new_node

    def addLast(self, value):
        new_node = SimpleNode(value)
        if self.isEmpty():
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def add(self, value, n):
        if n < 0:
            print("Debe ser un índice positivo")
            return
        if n == 0:
            self.addFirst(value)
            return
        new_node = SimpleNode(value)
        current = self.head
        for i in range(n - 1):
            if current is None:
                print("Fuera de rango")
                return
            current = current.next
        if current is None:
            print("Fuera de rango")
            return
        new_node.next = current.next
        current.next = new_node

    def removeFirst(self):
        if self.isEmpty():
            print("La lista está vacía")
            return
        removed_value = self.head.value
        self.head = self.head.next
        return removed_value

    def removeLast(self):
        if self.isEmpty():
            print("La lista está vacía")
            return
        if self.head.next is None:
            return self.removeFirst()
        current = self.head
        while current.next.next:
            current = current.next
        removed_value = current.next.value
        current.next = None
        return removed_value

    def remove(self, index):
        if self.isEmpty():
            print("La lista está vacía")
        if index < 0:
            print("Debe ser un índice positivo")
        if index == 0:
            return self.removeFirst()
        current = self.head
        for i in range(index - 1):
            if current.next is None:
                print("Fuera de rango")
                return
            current = current.next
        if current.next is None:
            print("Fuera de rango")
            return
        removed_value = current.next.value
        current.next = current.next.next
        return removed_value

    def search(self, value):
        current = self.head
        position = 0
        while current:
            if current.value == value:
                return position
            current = current.next
            position += 1

        print("No encontrado")
        return -1

    def get(self, index):
        if index < 0:
            print("Debe ser un índice positivo")
        current = self.head
        for i in range(index):
            if current is None:
                print("Fuera de rango")
            current = current.next
        if current is None:
            print("No encontrado")
        return current.value

    def toString(self):
        values = []
        current = self.head
        while current:
            values.append(repr(current.value))
            current = current.next
        return " -> ".join(values)


# Menú de opciones
def menu():
    print("\n################################")
    print("         Menú de opciones:")
    print("################################")
    print("1. Agregar al inicio")
    print("2. Agregar al final")
    print("3. Agregar en posición")
    print("4. Eliminar primero")
    print("5. Eliminar último")
    print("6. Eliminar en posición")
    print("7. Buscar")
    print("8. Obtener")
    print("9. Mostrar lista")
    print("0. Salir")
    return input("Seleccione una opción: ")

def main():
    lista = SimpleList()
    while True:
        opcion = menu()
        if opcion == "1":
            valor = input("Ingrese un valor: ")
            lista.addFirst(valor)
        elif opcion == "2":
            valor = input("Ingrese un valor: ")
            lista.addLast(valor)
        elif opcion == "3":
            valor = input("Ingrese un valor: ")
            posicion = int(input("Ingrese la posición: "))
            lista.add(valor, posicion)
        elif opcion == "4":
            lista.removeFirst()
        elif opcion == "5":
            lista.removeLast()
        elif opcion == "6":
            posicion = int(input("Ingrese la posición: "))
            lista.remove(posicion)
        elif opcion == "7":
            valor = input("Ingrese un valor: ")
            lista.search(valor)
        elif opcion == "8":
            posicion = int(input("Ingrese la posición: "))
            lista.get(posicion)
        elif opcion == "9":
            print(lista.toString())
        elif opcion == "0":
            break
        else:
            print("Opción no válida")

main()
```
