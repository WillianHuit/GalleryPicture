```
class BSTNode:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.left = None
        self.right = None
class BST:
    def __init__(self):
        self.root = None

    def insert(self, key, value):
        if self.root is None:
            self.root = BSTNode(key, value)
        else:
            self._insert_rec(self.root, key, value)

    def _insert_rec(self, node, key, value):
        if key < node.key:
            if node.left is None:
                node.left = BSTNode(key, value)
            else:
                self._insert_rec(node.left, key, value)
        elif key > node.key:
            if node.right is None:
                node.right = BSTNode(key, value)
            else:
                self._insert_rec(node.right, key, value)
        else:
            node.value = value  # Update existing key

    def search(self, key):
        return self._search_rec(self.root, key)

    def _search_rec(self, node, key):
        if node is None:
            return None
        if key == node.key:
            return node.value
        elif key < node.key:
            return self._search_rec(node.left, key)
        else:
            return self._search_rec(node.right, key)

    def search_by(self, predicate):
        results = []

        def _traverse(node):
            if node is None:
                return
            _traverse(node.left)
            if predicate(node.value):
                results.append(node.value)
            _traverse(node.right)

        _traverse(self.root)
        return results

  class Stack:
    def __init__(self):
        self._data = []

    def push(self, item):
        self._data.append(item)

    def pop(self):
        if self.is_empty():
            return None
        return self._data.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self._data[-1]

    def is_empty(self):
        return len(self._data) == 0

    def size(self):
        return len(self._data)

    def clear(self):
        self._data.clear()
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self.items[-1]

    def size(self):
        return len(self.items)

    def clear(self):
        self.items = []

class Queue:
    def __init__(self):
        self._data = []
        self._head = 0

    def enqueue(self, item):
        self._data.append(item)

    def dequeue(self):
        if self.is_empty():
            return None
        item = self._data[self._head]
        self._head += 1
        # Compactar ocasionalmente para evitar crecimiento de memoria
        if self._head > 50 and self._head > len(self._data) // 2:
            self._data = self._data[self._head:]
            self._head = 0
        return item

    def peek(self):
        if self.is_empty():
            return None
        return self._data[self._head]

    def is_empty(self):
        return self._head >= len(self._data)

    def size(self):
        return len(self._data) - self._head

    def clear(self):
        self._data = []
        self._head = 0


```

```
from bst import BST
from queue import Queue
from stack import Stack

class Ticket:
    _next_id = 1

    def __init__(self, title, description, priority, responsible, status="OPEN"):
        self.id = Ticket._next_id
        Ticket._next_id += 1
        self.title = title
        self.description = description
        self.priority = priority
        self.responsible = responsible
        self.status = status
        self.history = Stack()
        self.history.push(f"CREATED status={status}")

    def get_all_priorities(self):
        return ["BAJA", "MEDIA", "ALTA"]

    def get_all_statuses(self):
        return ["OPEN", "IN_PROGRESS", "CLOSED"]

    def update_status(self, new_status):
        old = self.status
        self.status = new_status
        self.history.push(f"STATUS {old} -> {new_status}")

    def __repr__(self):
        return (f"Ticket(id={self.id}, title={self.title!r}, priority={self.priority}, "
                f"responsible={self.responsible!r}, status={self.status})")

class TicketSystem:
    def __init__(self):
        self.queue = Queue()
        self.index = BST()

    def create_ticket(self, title, description, priority, responsible, status="OPEN"):
        ticket = Ticket(title, description, priority, responsible, status)
        # Enqueue ID for attention order
        self.queue.enqueue(ticket.id)
        # Index in BST by ID
        self.index.insert(ticket.id, ticket)
        return ticket

    def get_ticket(self, ticket_id):
        return self.index.search(ticket_id)

    def update_status(self, ticket_id, new_status):
        t = self.get_ticket(ticket_id)
        if not t:
            return None
        t.update_status(new_status)
        return t

    def next_ticket(self):

        ticket_id = self.queue.dequeue()
        if ticket_id is None:
            return None
        return self.get_ticket(ticket_id)

    def search_by_responsible(self, responsible):
        return self.index.search_by(lambda t: t.responsible == responsible)

    def search_by_priority(self, priority):
        return self.index.search_by(lambda t: t.priority == priority)

    def search_by_id(self, ticket_id):
        t = self.get_ticket(ticket_id)
        return t if t else None

    def all_tickets(self):
        return self.index.search_by(lambda t: True)


```

```
from ticket_system import TicketSystem

def create_ticket(ts):
    title = input("Ingresa el título del ticket: ")
    description = input("Ingresa descripción: ")
    priority = input("Ingresa la prioridad del ticket (BAJA, MEDIA, ALTA): ")
    while priority not in ["BAJA", "MEDIA", "ALTA"]:
        print("Prioridad inválida. Por favor ingresa BAJA, MEDIA o ALTA.")
        priority = input("Ingresa la prioridad del ticket (BAJA, MEDIA, ALTA): ")
    responsible = input("Ingresa la persona responsable: ")
    ticket = ts.create_ticket(title, description, priority, responsible)
    print(f"Ticket creado: {ticket}")

def get_ticket(ts):
    ticket_id = int(input("Ingresa el ID del ticket: "))
    ticket = ts.get_ticket(ticket_id)
    print(f"Detalles del Ticket: {ticket}")

def update_ticket_status(ts):
    ticket_id = int(input("Ingresa el ID del ticket: "))
    new_status = input("Ingresa el nuevo estado  (OPEN, IN_PROGRESS, CLOSED): ")
    while new_status not in ts.get_all_statuses():
        print("Estado inválido. Por favor ingresa OPEN, IN_PROGRESS o CLOSED.")
        new_status = input("Ingresa el nuevo estado  (OPEN, IN_PROGRESS, CLOSED): ")
    ticket = ts.update_status(ticket_id, new_status)
    print(f"Ticket actualizado: {ticket}")

def next_ticket(ts):
    ticket = ts.next_ticket()
    print(f"Siguiente Ticket: {ticket}")

def search_by_responsible(ts):
    responsible = input("Ingresa la persona responsable: ")
    tickets = ts.search_by_responsible(responsible)
    print(f"Tickets para {responsible}: {tickets}")

def search_by_priority(ts):
    priority = input("Ingresa la prioridad (BAJA, MEDIA, ALTA): ")
    tickets = ts.search_by_priority(priority)
    print(f"Tickets con prioridad {priority}: {tickets}")

def create_clear_lines():
    print("\n" * 35)

def main():
    MENU = """
=======================================
        Sistema de Tickets
=======================================
 1. Crear Ticket
 2. Obtener Ticket por ID
 3. Actualizar Estado del Ticket
 4. Siguiente Ticket
 5. Buscar Tickets por Responsable
 6. Buscar Tickets por Prioridad
 7. Listar Todos los Tickets
 0. Salir
=======================================
Ingresa una opción: """

    ts = TicketSystem()

    while True:
        choice = input(MENU).strip()
        print("")
        if choice == "1":
            create_clear_lines()
            create_ticket(ts)
        elif choice == "2":
            create_clear_lines()
            get_ticket(ts)
        elif choice == "3":
            create_clear_lines()
            update_ticket_status(ts)
        elif choice == "4":
            create_clear_lines()
            next_ticket(ts)
        elif choice == "5":
            create_clear_lines()
            search_by_responsible(ts)
        elif choice == "6":
            create_clear_lines()
            search_by_priority(ts)
        elif choice == "7":
            create_clear_lines()
            print("Tickets en el sistema:")
            tickets = ts.all_tickets()
            for ticket in tickets:
                print(f" - {ticket}")
        elif choice == "0":
            print("Saliendo del Sistema de Tickets.")
            break
        else:
            print("Opción Invalida, por favor intenta de nuevo.")



if __name__ == "__main__":
	main()
```
